```python
import matplotlib.pyplot as plt

# Count the gender values
gender_counts = df.value_counts()

# Create bar graph
gender_counts.plot(kind="bar", color="skyblue")

# Add labels and title
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Gender Distribution")
plt.xticks(rotation=0)

# Display graph
plt.show()
```

```python
import pandas as pd
import seaborn as sns

# Load dataset
df = pd.read_csv(r"C:\Users\brund\Downloads\online food delivery dataset.csv")

# Display first 5 rows
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Create scatterplot
sns.scatterplot(
    data=df,
    x="Age",
    y="Monthly Income",
    hue="Gender"
)

# Add title and labels
plt.title("Age vs Monthly Income")
plt.xlabel("Age")
plt.ylabel("Monthly Income")

# Display plot
plt.show()
```

```python
import matplotlib.pyplot as plt

sns.boxplot(
    data=df,
    x="Gender",
    y="Monthly Income"
)

plt.title("Monthly Income by Gender")
plt.xlabel("Gender")
plt.ylabel("Monthly Income")

plt.show()
```
